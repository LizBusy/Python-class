Abstraction in Python means:

“I will create a rule/template, and force other classes to follow it.”

--How to Achieve Abstraction in Python

Unlike Java, Python does not have built-in abstract classes by default. We must import the abc module (Abstract Base Classes).

The Three Rules of Abstraction:

1. Inheritance: You must inherit from ABC.
2. Decorator: You must use @abstractmethod on methods you want to enforce.
3. No Instantiation: You cannot create an object from an Abstract Class. If you try, Python will throw an error.


# How to do it in Python

```python
# Example 1: The Automobile Standard

# In this example, Car is an abstract concept. You cannot buy a generic "Car"; you must buy a specific brand

# Import the necessary tools
from abc import ABC, abstractmethod 

# 2. Define the Abstract Base Class
class Car(ABC):
    @abstractmethod 
    def mileage(self): # This acts as a strict rule. 
        pass           # Any subclass MUST implement this method, or Python will crash. 
                                    
                                    
# Create concrete classses
class Tesla(Car):
    def mileage(self):
        print("The mileage is 30kmph (Electric)")

class Suzuki(Car):
    def mileage(self):
        print("The mileage is 25kmph (Petrol)")

class Duster(Car):
    def mileage(self):
        print("The mileage is 24kmph (Diesel)")

class Renault(Car):
    def mileage(self):
        print("The mileage is 27kmph (Hybrid)") 

## Execution
#  Trying to create the abstract class directly will FAIL
c = Car() 
# Error: Can't instantiate abstract class Car with abstract method mileage

# Creating Concrete Objects
t = Tesla()
s = Suzuki()
d = Duster()
r = Renault()

# Calling the methods

t.mileage()
s.mileage()
d.mileage()
r.mileage()
```

```python
# Step 1: Start with a normal class (no abstraction)

class Dog:
    def sound(self):
        print("Dog barks") 

d = Dog()
d.sound()

# This is just a regular class- has no abstraction or rules


# Step 2: Now imagine we want a RULE

#“Every animal MUST have a sound() method”

#Instead of repeating this in every class, we create a base (abstract) class

#Step 3: 
from abc import ABC, abstractmethod

# ABC = tells python this is an abstract class
# abstractmethod = marks a method as a rule

#Step 4: Create the abstract class(the rule)

class Animal(ABC) :
   @abstractmethod
   def sound(self) :
      pass          # Any class that is an animal must have a sound()
                          

#Step 5: Implementation
class Dog(Animal):
   def sound(self) :
      print("Bark")

class Cat(Animal) :
   def sound(self) :
      print("meow")

class Horse(Animal):
   def sound(self):
      print("neigh")


#Step 7: Use the classes   
d = Dog()
c = Cat()
h = Horse()

d.sound()  #Bark
c.sound()  #Meow
h.sound()  #Neigh
```

```python
from abc import ABC, abstractmethod

#1 The Abstract Base Class
class Polygon(ABC) :
    @abstractmethod
    def sides(self):
        """Every polygon must define how mayny sides it has."""
        pass

#2. concrete classes
class Triangle(Polygon):
    def sides(self):
        print("Triangle has 3 sides")

class Pentagon(Polygon):
    def sides(self):
        print("Pentagon has 5 sides")

class Hexagon(Polygon):
    def sides(self):
        print("Hexagon has 6 sides")

class Square(Polygon):
    def sides(self):
        print("Square has 4 sides")

# EXECUTION

# Creating objects
t = Triangle()
s = Square()
p = Pentagon()
k = Hexagon()

# Invoking the common interface
t.sides()
s.sides()
p.sides()
k.sides()
```

```python
# Inheritance - Reusing and Extending

#Inheritance lets a child reuese everything from a parent class andor override as needed.

#This enforces the DRYprinciple(Don't repeat yourself).

#PARENT(BASE) CLASS

class Animal:
    def __init__(self, name, age):
        self.name = name
        self.age = age
        self.is_alive = True
 
    def eat(self):
        print(f"{self.name} is eating")

    def sleep(self):
        print(f"{self.name} is sleeping.")

    def __str__(self):
        return f"{self.__class__.__name__}: {self.name}, Age{self.age}"

        #CHILD CLASS
class Dog(Animal) : #Dog inherits everything from Animal
    def __init__(self, name, age, breed):
        super().__init__(name, age)   #callparent's __init__ first
        self.breed = breed            # Add specific attribute
    
    def bark(self) :
        print(f"{self.name} says: Woof")

    def fetch(self, item):
        print(f"{self.name} fetches the {item}! ")

class Cat(Animal):
    def __init__(self, name, age, is_indoor):
        super().__init__(name, age)
        self.is_indoor = is_indoor

    def purr(self):
        print(f"{self.name} purrs contentedly.")

    def sleep(self):
        super().sleep()
        print(f" ...{self.name} sleeps for 16 hours straight. ")

# MULTI-LEVEL INHERITANCE
class GuideDog(Dog):    #GuideDog inherits from Dog which inherits from Animal
    def __init__(self, name, age, breed, owner):
        super().__init__(name, age, breed)
        self.owner = owner

    def guide(self):
        print(f"{self.name} carefully guides {self.owner} across the street. ")

 #TESTING
 #                        



```

ENCAPSULATION
Hiding data and controlling access to it inside a class- Contolling how it's accessed using methods



```python
# Public variables- Cna be accessed anywhere/ modified anywhere
#no restrictions

# You can use s1.name to change names from Bianca to Stefan to Joy directly.

class Student:
    def __init__ (self, name):
        self.name = name
s1 = Student("Bianca")
print(s1.name)
s1.name = "Stefan"
print(s1.name)
s1.name = "Joy"
print(s1.name)
```

```python
# Protected variables - still accessible - changeable - for internal use

class Employee:
    def __init__(self):
        self._salary = 5000
emp = Employee()
```

```python
class Student:
    def __init__(self):
        self.__grade = 0
    def set_grade(self, grade):
        if 0 <= grade <= 100:
            self.__grade = grade  #__grade is hidden internally inside set_grade()
            # and set_garde() contains my conditional (if) statement
        else:
            print("Invalid Grade")

  # get grade
    def get_grade(self):
        return self.__grade

# Create object = s1 
s1 = Student() # fromm class Student
s1.set_grade(85)

# Modification
s1.__grade = 1000
print(s1.get_grade())
```

```python
# Without Enscapulation

class Student:
    def __init__(self):
        self.grade = 0
    def set_grade(self, grade):
        if 0 <= grade <= 100:
            self.grade = grade   # __grade is hidden internally inside
        else:
            print("Invalid Grade")

# create object = s1
s1 = Student()
#method using the object
s1.set_grade(85)
print(s1.grade)

#Modification
s1.grade = 1000 # Not valid grade but python still prints it because its a public variable
print(s1.grade)
```

```python
class Bank_account:
    def __init__(self, balance):
        self.__balance = balance  # Private variable enscapulation
    
    def deposit(self, amount):
        self.__balance += amount

    def withdraw(self, amount):
        if amount <= self.__balance:
            self.__balance -= amount
        else:
            print("Insufficient funds")

            #GET METHOD
    def get_balance(self):
        return self.__balance

acc = Bank_account(1000)
acc.deposit(500)
acc.withdraw(300)
print(acc.get_balance())        
```
