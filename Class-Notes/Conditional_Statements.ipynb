# Conditional Statements

Conditional statements are statements that perform a specific task if the conditon is 'True' and if it is 'False' then do nothing.

'if' > condition is met > perform action

```python
# As a vendor at a concert you have a target number of units you should sell

sales_target = 300
units_sold = 279

# Let me compare the sales

if units_sold >= sales_target:
   print("Target Achieved")

# Check if we were close to the target
elif units_sold >= 250:
   print("Target almost achieved")

#Otherwise...
elif units_sold == 290:
   print("10 units left to hit target")

elif units_sold == 279:
   print("279 units sold, 21 units left to hit target")

else:
   print("Target not achieved")
```

```python
x = 5
y = 10

if x < y:
    print('x is less than y')
elif x > y:
    print('x is greater than y')
else:
    print('x and y are equal')
```

```python
if x == y:
    print('x and y are equal')
else:
    if x < y:
        print('x is less than y')
    else:
        print('x is greater than y')
```

```python
# Instead of nested if:
if 0 < x and x < 10:
    print('x is a positive single-digit number.')
```

```python
x = 6
y = 0
# The 'y != 0' guard prevents the division-by-zero crash
if x >= 2 and y != 0 and (x / y) > 2:
    print("Passed")
else:
    print("Safe")
```
