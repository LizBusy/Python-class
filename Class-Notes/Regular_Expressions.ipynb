### 'PATTERN' - 'MEANING'

'\d'  - digits

'\D'  - not a digit

'\w'  - letters and numbers

'\W'  - not a word character

'\s'  - space characters

'\S'  - non-space characters

'.'   - any character

'+'   - 1 or more characters

`*`     - 0 or more characters

'?'     - optional

'^'     - start

'$'     - end

1.Search complex patterns

```python
# Working with structured text eg. emails, numbers,logs
# re.search() - Search for a pattern in a string and return the first match

import re

text = "My number is 0792292221"
result = re.search(r"\d{10}" , text)

if result:
    print("Phone number found" , 
          result.group())
```

2. Extract Data

```python
# When Extracting data we can use re.findall() to find all accurrences of pattern in a string

import re
text = "Emails : k@example.com, l@example.com"
emails = re.findall(r"\S+@\S+", text)
print("Emails found:" , text)
```

3. Replace Patterns

```python
# When replacing Text we can use re.sub() to replace patterns with the new text
import re

text = "Call me at 0792292221"
new_text = re.sub(r"\d" , "*" , text)
print(new_text)

text = "Call me at 0792292221"
new_text = re.sub(r"07(\d{8})", r"07********" , text)
print(new_text)

text = "Call me at 0792292221"
new_text = re.sub(r"(07\d{2})(\d{4}\d{2})" , r"\1****\2" , text)
print(new_text)
```

4. Validate Input

```python
# re.match() - Check if the beginning of a string matches a pattern
# # It returns a match if the pattern matcehs at the start of the string, otherwise it returns None

import re

text = "Hello, World!"
match = re.match(r"Hello", text)
if match:
    print("Match found:" , match.group())
else:
    print("None")
```
