Data Structures in Python

Lists in Python

- Lists store multiple values in a single variable
  - Lists can contain any combination of data type eg boolean, string, int

- Lists are ordered
  - One can use subsetting or indexing

```python
fuel_prices = [90, 150, 100, 194, 60, 76]
print(fuel_prices)
```

#### Accessing Elements of a list
- We use the

```python
print(fuel_prices)
```

```python
(fuel_prices)[2]
```

Slicing in Lists / Accessing Multiple Elements in a List

```python
fuel_prices[2:5]
```

```python
# Get the first three fuel prices
# These are values from index zero to index two (not inclusive of three)
fuel_prices[:3]
```

```python
# Access all the elements from the fourth index onwards

fuel_prices[3:]
```

##### Alternating Access

```python
print(fuel_prices)
```

```python
# Accessing Every other element in the List
fuel_prices[::2]
```

```python
# Access every 3rd element starting from the second index element(index 1)
fuel_prices[1::3]
```

#### Sets in Python

- Contain unique Data
- They are unchangeable
  - You can add or remove but you can't change what's already there

A Set is defined by = {} 

```python
mokimo_ingredients = {"Irish potatoes", "Pumpkin leaves", "Margerine", "Salt",
                       "Beans", "Peas", "Onions", "Carrots", "Pepper"}
print(mokimo_ingredients)
```

```python
# Changing a set to be a list
mokimo_ingredients_list = ["Irish potatoes", "Pumpkin leaves", "Margerine", "Salt",
                       "Beans", "Peas", "Onions", "Carrots", "Pepper"]
print(mokimo_ingredients_list)
```

```python
# Changing a list to a set
mokimo_ingredients_list = set(mokimo_ingredients_list)
print(mokimo_ingredients_list)
```

```python
print(type(mokimo_ingredients_list))
```

```python
print(mokimo_ingredients_list)
```

```python
# Sorting my set
sorted_mokimo_ingredients = sorted(mokimo_ingredients_list)
print(sorted_mokimo_ingredients)
```

```python
print(type(sorted_mokimo_ingredients))
```
