url = https://webscraper.io/test-sites/e-commerce/allinone/phones

#to mask as a browser

```python

#masking to look like a normal user using window and browser, not look like a bot
header = {
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/115.0.0.0 Safari/537.36"}


# get request is how you ask for data from a url
# response stores what the server sends back
response = requests.get(url, headers = header)

#python sees the html

soup = BeautifulSoup(response.content, "html.parser")

#finding the elements , .find() nd .findall()
#.find(arguments)

product = soup.find("div" , class_ = "caption")

#we want the prices and product name but we use 
# div because it has both product name and price

#we use a loop to get the product and the price

for productinfo in product : # loops through the div tag
    
    productname = product.find("a" , class_ ="title") # we are checking for phone name <a>

    cleanname = productname.text.strip() #removes the tags and spaces

    productprice = product.find("span" , itemprop = "price") # phone price tag <span>

    cleanprice = productprice.text.strip() #same cleaning as name

print(f' the product, {cleanname}, price is {cleanprice}')
```

```python
#now we want to get all items with name and prices
#to find all items in the page with prices

allproducts = soup.find_all("div", class_="caption")

#loop to go over all items


for info in allproducts :
    name = info.find("a" , class_ = "title")
    nameclean = name.text.strip()
    print(nameclean)
    prices = info.find("span" , itemprop = "price") 
    pricesclean = prices.text.strip()
    print(f"{nameclean} | {pricesclean}")
```
